2. В якому місті знайходиться філіал з найвищим прибутком? Може в тому місці нам розмітисти ще один магазин.  
Знайдіть відповідь за допомогою функціоналу Pandas.

```python
summary = (
    df.groupby(["Branch", "City"])
    .agg(
        gross_income=("gross income", "sum"),
        invoices=("Invoice ID", "count"),
        avg_check=("Total", "mean"),
    )
    .assign(income_share_pct=lambda t: t["gross_income"] / t["gross_income"].sum() * 100)
    .round(2)
    .sort_values("gross_income", ascending=False)
)

print(summary)
```

Кожен філіал знаходиться в одному місті. Найбільший валовий дохід у філіалі С, місто Naypyitaw, при найменшій кількості інвойсів. При цьому, середній чек тут найбільший, а доля валового доходу більша на 1,36 від інших філіалів. Згідно цих даних не доцільно розміщати ще один магазин.

Гіпотеза 1.
Кількість інвойсів та середній чек за типом покупців.
Висновок та рекомендації для бізнесу:

Ефективність програми лояльності у філіалах A (Yangon) та B (Mandalay): Учасники програми лояльності (Member) витрачають у середньому більше за одну покупку ($321.18 та $325.48 проти $303.83 та $314.33 у Normal), хоча кількість їхніх візитів трохи менша. Для цих магазинів доцільно активніше стимулювати звичайних покупців (Normal) до реєстрації в програмі лояльності (наприклад, пропонувати миттєву знижку або бонусні бали на касі при оформленні картки), щоб конвертувати їх у сегмент із вищим середнім чеком.

Особливість філіалу C (Naypyitaw): Тут спостерігається найбільша кількість постійних клієнтів (Member — 169 чеків), проте середній чек у них ($336.58) майже не відрізняється від звичайних покупців ($337.66). Це свідчить про те, що поточна програма лояльності в цьому місті добре утримує клієнтів, але не мотивує їх купувати більше за один візит. Для філіалу C варто впровадити механіки збільшення середнього чека саме для власників карток (наприклад, персоналізовані пропозиції, кешбек від певної суми чека або акції формату «купи на $350 — отримай подвійні бонуси»).

Оскільки розподіл покупок між Member та Normal у всіх магазинах близький до 50/50, мережа має значний потенціал для росту доходу через переведення половини наявної клієнтської бази в програму лояльності з подальшою персоналізацією пропозицій.

```python
table_customer_type = pd.pivot_table(
    df,
    index=['Branch', 'City'],
    columns='Customer type',
    values=['Invoice ID', 'Total'],
    aggfunc={'Invoice ID': 'count', 'Total': 'mean'},
).round(2)

print(table_customer_type)
```

Гіпотеза 2.
Найбільш продаваємі категорії товару по філіям (містам).

Висновок та рекомендації для бізнесу:

Локалізація асортименту та управління запасами (мерчандайзинг): Попит на категорії суттєво відрізняється залежно від міста, тому мережі варто відмовитися від однакового розподілу товарів між магазинами на користь локального підходу:

Філіал A (Yangon): абсолютний лідер за виручкою та кількістю — Home and lifestyle ($22,417.20 / 371 од.), а найслабша категорія — Health and beauty ($12,597.75). Варто розширити площу та асортимент товарів для дому, а для категорії краси й здоров'я провести промоакції або переглянути асортиментну матрицю.

Філіал B (Mandalay): найкраще продаються Sports and travel ($19,988.20) та Health and beauty ($19,980.66), тоді як Food and beverages має найнижчі показники ($15,214.89). Магазину доцільно робити акцент на просуванні товарів для спорту, подорожей і догляду, а також перевірити свіжість/асортимент продуктового відділу.

Філіал C (Naypyitaw): беззаперечні лідери — Food and beverages ($23,766.86 / 369 од.) та Fashion accessories ($21,560.07), тоді як Home and lifestyle (лідер у Yangon) тут посідає останнє місце ($13,895.55). Рекомендується збільшити закупівлі продуктів харчування та модних аксесуарів, скоротивши надлишкові складські запаси товарів для дому.

Оптимізація маркетингових бюджетів (таргетована реклама): Замість загальних рекламних кампаній на всю мережу варто запускати локальну рекламу, підсилюючи топові категорії кожного міста (наприклад, реклама продуктів та аксесуарів у Naypyitaw, а товарів для дому — у Yangon), щоб максимізувати повернення інвестицій (ROI).

```python
table_customer_type2 = pd.pivot_table(
    df,
    index=['Branch', 'City', 'Product line'],
    values=['Quantity', 'Total'],
    aggfunc={'Quantity': 'sum', 'Total': 'sum'},
).round(2).sort_values(by=['Branch', 'Total'], ascending=[True, False])

print(table_customer_type2)
```
